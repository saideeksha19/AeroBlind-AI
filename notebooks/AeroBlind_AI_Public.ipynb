```python

import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error

TARGETS = ["t_ch4", "t_no2", "t_co", "t_aai", "t_firms"]

# Exclude identifiers and all target columns
FEATURES = [c for c in test.columns if c != "sample_id"]

X = train[FEATURES].copy()
groups = train["site_key"]

# Five folds, with sites kept separate
gkf = GroupKFold(n_splits=5)

results = []

print("=" * 65)
print("AEROBLIND AI - EXPERIMENT 01")
print("SITE-GROUPED TABULAR BASELINE")
print("=" * 65)

for target in TARGETS:

    print(f"\nTraining target: {target}")

    y = train[target]
    oof_extra = np.full(len(train), np.nan)
    oof_dummy = np.full(len(train), np.nan)

    fold_scores = []

    for fold, (train_idx, valid_idx) in enumerate(
        gkf.split(X, y, groups), start=1
    ):

        # Use only rows with available labels
        valid_train_idx = train_idx[y.iloc[train_idx].notna().values]
        valid_valid_idx = valid_idx[y.iloc[valid_idx].notna().values]

        if len(valid_train_idx) == 0 or len(valid_valid_idx) == 0:
            print(f"Fold {fold}: skipped (no available labels)")
            continue

        X_train = X.iloc[valid_train_idx]
        y_train = y.iloc[valid_train_idx]

        X_valid = X.iloc[valid_valid_idx]
        y_valid = y.iloc[valid_valid_idx]

        # Reference: training-fold mean prediction
        dummy = DummyRegressor(strategy="mean")
        dummy.fit(X_train, y_train)

        dummy_pred = dummy.predict(X_valid)

        # Extra Trees model
        model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("regressor", ExtraTreesRegressor(
                n_estimators=300,
                max_features=0.8,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            ))
        ])

        model.fit(X_train, y_train)
        predictions = model.predict(X_valid)

        oof_extra[valid_valid_idx] = predictions
        oof_dummy[valid_valid_idx] = dummy_pred

        extra_rmse = np.sqrt(
            mean_squared_error(y_valid, predictions)
        )

        dummy_rmse = np.sqrt(
            mean_squared_error(y_valid, dummy_pred)
        )

        fold_scores.append(extra_rmse)

        print(
            f"Fold {fold}: "
            f"ExtraTrees RMSE={extra_rmse:.5f} | "
            f"Mean baseline RMSE={dummy_rmse:.5f} | "
            f"Validation labels={len(y_valid)}"
        )

    # Overall out-of-fold performance
    mask = y.notna().values & np.isfinite(oof_extra)

    extra_oof_rmse = np.sqrt(
        mean_squared_error(y[mask], oof_extra[mask])
    )

    dummy_oof_rmse = np.sqrt(
        mean_squared_error(y[mask], oof_dummy[mask])
    )

    results.append({
        "Target": target,
        "Valid labels": int(mask.sum()),
        "ExtraTrees OOF RMSE": extra_oof_rmse,
        "Mean baseline OOF RMSE": dummy_oof_rmse,
        "Fold RMSE mean": np.mean(fold_scores)
    })

    print(f"Overall ExtraTrees OOF RMSE: {extra_oof_rmse:.5f}")
    print(f"Overall Mean baseline RMSE: {dummy_oof_rmse:.5f}")

# Final comparison
results_df = pd.DataFrame(results)

print("\n" + "=" * 65)
print("FINAL BASELINE RESULTS")
print("=" * 65)

display(results_df.round(5))

print(
    "\nEXTRATREES MCRMSE:",
    round(results_df["ExtraTrees OOF RMSE"].mean(), 5)
)

print(
    "MEAN BASELINE MCRMSE:",
    round(results_df["Mean baseline OOF RMSE"].mean(), 5)
)

print("\nEXPERIMENT 01 COMPLETE")
```

```python

import numpy as np
import pandas as pd

from catboost import CatBoostRegressor
from sklearn.model_selection import GroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error

TARGETS = ["t_ch4", "t_no2", "t_co", "t_aai", "t_firms"]
FEATURES = [c for c in test.columns if c != "sample_id"]

X = train[FEATURES].copy()
groups = train["site_key"]

gkf = GroupKFold(n_splits=5)

all_results = []

print("=" * 70)
print("AEROBLIND AI - EXPERIMENT 02")
print("CATBOOST vs EXTRATREES vs MEAN BASELINE")
print("=" * 70)

for target in TARGETS:

    print(f"\n{'=' * 55}")
    print("TARGET:", target)
    print("=" * 55)

    y = train[target]

    oof_cat = np.full(len(train), np.nan)
    oof_extra = np.full(len(train), np.nan)
    oof_dummy = np.full(len(train), np.nan)

    for fold, (train_idx, valid_idx) in enumerate(
        gkf.split(X, y, groups), start=1
    ):

        # Remove rows with missing target labels
        tr_idx = train_idx[y.iloc[train_idx].notna().values]
        va_idx = valid_idx[y.iloc[valid_idx].notna().values]

        if len(tr_idx) == 0 or len(va_idx) == 0:
            print(f"Fold {fold}: skipped")
            continue

        X_train = X.iloc[tr_idx]
        y_train = y.iloc[tr_idx]

        X_valid = X.iloc[va_idx]
        y_valid = y.iloc[va_idx]

        # Mean baseline
        dummy = DummyRegressor(strategy="mean")
        dummy.fit(X_train, y_train)
        pred_dummy = dummy.predict(X_valid)

        # Extra Trees
        extra_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("regressor", ExtraTreesRegressor(
                n_estimators=300,
                max_features=0.8,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            ))
        ])

        extra_model.fit(X_train, y_train)
        pred_extra = extra_model.predict(X_valid)

        # CatBoost
        cat_model = CatBoostRegressor(
            iterations=700,
            depth=5,
            learning_rate=0.03,
            loss_function="RMSE",
            l2_leaf_reg=5,
            random_seed=42,
            verbose=False,
            allow_writing_files=False,
            thread_count=-1
        )

        cat_model.fit(X_train, y_train)
        pred_cat = cat_model.predict(X_valid)

        # Store out-of-fold predictions
        oof_dummy[va_idx] = pred_dummy
        oof_extra[va_idx] = pred_extra
        oof_cat[va_idx] = pred_cat

        # Fold metrics
        rmse_dummy = np.sqrt(mean_squared_error(y_valid, pred_dummy))
        rmse_extra = np.sqrt(mean_squared_error(y_valid, pred_extra))
        rmse_cat = np.sqrt(mean_squared_error(y_valid, pred_cat))

        print(
            f"Fold {fold} | "
            f"Mean: {rmse_dummy:.5f} | "
            f"ExtraTrees: {rmse_extra:.5f} | "
            f"CatBoost: {rmse_cat:.5f}"
        )

    # Overall OOF metrics
    mask = y.notna().values & np.isfinite(oof_cat)

    rmse_dummy = np.sqrt(
        mean_squared_error(y[mask], oof_dummy[mask])
    )

    rmse_extra = np.sqrt(
        mean_squared_error(y[mask], oof_extra[mask])
    )

    rmse_cat = np.sqrt(
        mean_squared_error(y[mask], oof_cat[mask])
    )

    all_results.append({
        "Target": target,
        "Valid labels": int(mask.sum()),
        "Mean baseline": rmse_dummy,
        "ExtraTrees": rmse_extra,
        "CatBoost": rmse_cat
    })

    print("\nOverall OOF:")
    print("Mean baseline:", round(rmse_dummy, 5))
    print("ExtraTrees:", round(rmse_extra, 5))
    print("CatBoost:", round(rmse_cat, 5))

# Final comparison
results_df = pd.DataFrame(all_results)

print("\n" + "=" * 70)
print("EXPERIMENT 02 - FINAL RESULTS")
print("=" * 70)

display(results_df.round(5))

print("\nOVERALL MCRMSE")
print("Mean baseline:", round(results_df["Mean baseline"].mean(), 5))
print("ExtraTrees:", round(results_df["ExtraTrees"].mean(), 5))
print("CatBoost:", round(results_df["CatBoost"].mean(), 5))

print("\nEXPERIMENT 02 COMPLETE")
```

```python

from pathlib import Path
import numpy as np
import rasterio
from collections import Counter

print("=" * 65)
print("AEROBLIND AI - EXPERIMENT 04")
print("SATELLITE IMAGE MAPPING AND QUALITY AUDIT")
print("=" * 65)

image_dir = DATA_DIR / "images"

# Build image lookup using filename stems
image_files = list(image_dir.glob("*.tif"))
image_lookup = {p.stem: p for p in image_files}

train_ids = set(train["sample_id"].astype(str))
test_ids = set(test["sample_id"].astype(str))
all_ids = train_ids | test_ids

# Check mapping coverage
train_missing = train_ids - image_lookup.keys()
test_missing = test_ids - image_lookup.keys()
extra_images = image_lookup.keys() - all_ids

print("\nIMAGE MAPPING")
print("Total TIFF images:", len(image_files))
print("Total unique train/test IDs:", len(all_ids))
print("Training IDs without images:", len(train_missing))
print("Test IDs without images:", len(test_missing))
print("Images without matching sample IDs:", len(extra_images))

# Inspect a sample of images
print("\nIMAGE QUALITY SAMPLE")

sample_ids = list(train["sample_id"].astype(str).head(10))

scl_counts = Counter()
cloud_means = []
reflectance_means = []

for sample_id in sample_ids:
    path = image_lookup.get(sample_id)

    if path is None:
        continue

    with rasterio.open(path) as src:
        arr = src.read()

        # Bands 1-12: spectral reflectance
        reflectance = arr[:12].astype(np.float32)
        reflectance_means.append(reflectance.mean())

        # Band 13: SCL categorical values
        scl = arr[12]
        scl_counts.update(
            Counter(scl.astype(int).ravel())
        )

        # Band 14: cloud probability
        cloud = arr[13].astype(np.float32)
        cloud_means.append(cloud.mean())

print("Images successfully inspected:", len(cloud_means))

if reflectance_means:
    print("Average spectral value:", round(
        float(np.mean(reflectance_means)), 3
    ))

if cloud_means:
    print("Average cloud probability:", round(
        float(np.mean(cloud_means)), 3
    ))

print("\nSCL CLASS PIXEL COUNTS")
for class_id, count in sorted(scl_counts.items()):
    print(f"Class {class_id}: {count}")

print("\n" + "=" * 65)
if not train_missing and not test_missing:
    print("SUCCESS: Every sample has a matching satellite image.")
else:
    print("WARNING: Some sample IDs have no matching image.")

print("EXPERIMENT 04 COMPLETE")
print("=" * 65)
```

```python

import numpy as np
import pandas as pd
import rasterio
from pathlib import Path
import time

print("=" * 70)
print("AEROBLIND AI - EXPERIMENT 05")
print("MULTIMODAL SATELLITE FEATURE EXTRACTION")
print("=" * 70)

image_dir = DATA_DIR / "images"
image_files = sorted(image_dir.glob("*.tif"))

records = []
start_time = time.time()

SCL_CLASSES = list(range(12))

for i, image_path in enumerate(image_files, start=1):

    with rasterio.open(image_path) as src:
        arr = src.read()

    row = {"sample_id": image_path.stem}

    # ------------------------------------------------
    # 1. SPECTRAL FEATURES: BANDS 1-12
    # ------------------------------------------------
    for band_idx in range(12):

        band = arr[band_idx].astype(np.float32)

        prefix = f"img_b{band_idx + 1}"

        row[f"{prefix}_mean"] = float(np.mean(band))
        row[f"{prefix}_std"] = float(np.std(band))
        row[f"{prefix}_p10"] = float(np.percentile(band, 10))
        row[f"{prefix}_p90"] = float(np.percentile(band, 90))

    # ------------------------------------------------
    # 2. SCENE CLASSIFICATION LAYER: BAND 13
    # ------------------------------------------------
    scl = arr[12].astype(np.uint8)
    scl_total = scl.size

    for class_id in SCL_CLASSES:
        row[f"scl_class_{class_id}_frac"] = float(
            np.count_nonzero(scl == class_id) / scl_total
        )

    # ------------------------------------------------
    # 3. CLOUD PROBABILITY: BAND 14
    # ------------------------------------------------
    cloud = arr[13].astype(np.float32)

    row["cloud_mean"] = float(np.mean(cloud))
    row["cloud_std"] = float(np.std(cloud))
    row["cloud_p90"] = float(np.percentile(cloud, 90))

    for threshold in [20, 40, 60, 80]:
        row[f"cloud_frac_above_{threshold}"] = float(
            np.mean(cloud > threshold)
        )

    records.append(row)

    if i % 250 == 0 or i == len(image_files):
        elapsed = time.time() - start_time

        print(
            f"Processed {i}/{len(image_files)} images | "
            f"Elapsed: {elapsed / 60:.1f} minutes"
        )

# ------------------------------------------------
# 4. CREATE FEATURE TABLE
# ------------------------------------------------
image_features = pd.DataFrame(records)

print("\n" + "=" * 70)
print("FEATURE EXTRACTION SUMMARY")
print("=" * 70)

print("Images processed:", len(image_features))
print("Image feature columns:", len(image_features.columns) - 1)
print("Feature table shape:", image_features.shape)

# Check duplicate IDs
print(
    "Duplicate image IDs:",
    image_features["sample_id"].duplicated().sum()
)

# Check missing feature values
feature_cols = [
    c for c in image_features.columns
    if c != "sample_id"
]

print(
    "Missing feature values:",
    image_features[feature_cols].isna().sum().sum()
)

# ------------------------------------------------
# 5. JOIN WITH TRAIN AND TEST
# ------------------------------------------------
train_multi = train.merge(
    image_features,
    on="sample_id",
    how="left",
    validate="one_to_one"
)

test_multi = test.merge(
    image_features,
    on="sample_id",
    how="left",
    validate="one_to_one"
)

print("\nMULTIMODAL DATASET SHAPES")
print("Training:", train_multi.shape)
print("Testing:", test_multi.shape)

print(
    "Missing training image features:",
    train_multi[feature_cols].isna().sum().sum()
)

print(
    "Missing testing image features:",
    test_multi[feature_cols].isna().sum().sum()
)

# ------------------------------------------------
# 6. SAVE FEATURES FOR REUSE
# ------------------------------------------------
output_path = Path("/kaggle/working/aeroblind_image_features.csv")

image_features.to_csv(output_path, index=False)

print("\nSaved image features:", output_path)

print("\nFEATURE SAMPLE")
display(image_features.head())

print("\nEXPERIMENT 05 COMPLETE")
print("=" * 70)
```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import GroupKFold
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_squared_error
from catboost import CatBoostRegressor
import time

print("=" * 70)
print("AEROBLIND AI - EXPERIMENT 06")
print("TABULAR vs MULTIMODAL VALIDATION")
print("=" * 70)

# Define columns
TARGETS = ["t_ch4", "t_no2", "t_co", "t_aai", "t_firms"]

TAB_FEATURES = [
    c for c in train.columns
    if c not in ["sample_id", "site_key"] + TARGETS
]

IMG_FEATURES = [
    c for c in image_features.columns if c != "sample_id"
]

MULTI_FEATURES = TAB_FEATURES + IMG_FEATURES

groups = train["site_key"]
cv = GroupKFold(n_splits=5)

results = []
start_time = time.time()

for target in TARGETS:
    print(f"\nTARGET: {target}")

    y_all = train[target]
    valid_label_mask = y_all.notna()

    y = y_all[valid_label_mask].astype(float)
    target_groups = groups[valid_label_mask]

    X_tab = train.loc[valid_label_mask, TAB_FEATURES].copy()
    X_multi = train_multi.loc[valid_label_mask, MULTI_FEATURES].copy()

    # Same site-grouped folds for both feature sets
    tab_oof = np.full(len(y), np.nan)
    multi_oof = np.full(len(y), np.nan)

    for fold, (tr_idx, val_idx) in enumerate(
        cv.split(X_tab, y, groups=target_groups), start=1
    ):
        y_tr = y.iloc[tr_idx]

        # ExtraTrees: tabular-only
        et_tab = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", ExtraTreesRegressor(
                n_estimators=300,
                max_features=0.8,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            ))
        ])
        et_tab.fit(X_tab.iloc[tr_idx], y_tr)
        tab_oof[val_idx] = et_tab.predict(X_tab.iloc[val_idx])

        # ExtraTrees: tabular + image features
        et_multi = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", ExtraTreesRegressor(
                n_estimators=300,
                max_features=0.8,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            ))
        ])
        et_multi.fit(X_multi.iloc[tr_idx], y_tr)
        multi_oof[val_idx] = et_multi.predict(X_multi.iloc[val_idx])

        print(f"  Fold {fold}/5 completed")

    tab_rmse = np.sqrt(mean_squared_error(y, tab_oof))
    multi_rmse = np.sqrt(mean_squared_error(y, multi_oof))

    improvement = (tab_rmse - multi_rmse) / tab_rmse * 100

    results.append({
        "target": target,
        "tabular_rmse": tab_rmse,
        "multimodal_rmse": multi_rmse,
        "improvement_percent": improvement
    })

    print(f"  Tabular RMSE:   {tab_rmse:.5f}")
    print(f"  Multimodal RMSE:{multi_rmse:.5f}")
    print(f"  Improvement:    {improvement:+.2f}%")

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("EXPERIMENT 06 RESULTS")
print("=" * 70)
display(results_df.round(5))

print("\nTABULAR MCRMSE:  ", round(results_df["tabular_rmse"].mean(), 5))
print("MULTIMODAL MCRMSE:", round(results_df["multimodal_rmse"].mean(), 5))

overall_improvement = (
    results_df["tabular_rmse"].mean()
    - results_df["multimodal_rmse"].mean()
) / results_df["tabular_rmse"].mean() * 100

print(f"OVERALL IMPROVEMENT: {overall_improvement:+.2f}%")

results_df.to_csv(
    "/kaggle/working/aeroblind_experiment06_results.csv",
    index=False
)

print("\nResults saved: aeroblind_experiment06_results.csv")
print(f"Elapsed time: {(time.time() - start_time) / 60:.1f} minutes")
```
